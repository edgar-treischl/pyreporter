```python
import pandas as pd

from pyreporter.utils import get_metadata, get_sname, get_plotdata
from pyreporter.meta_repository import MetaRepository

from pyreporter.limer import limer_connect, limer_list_surveys, limer_responses, limer_release, limer_n, limer_SIDs
```

```python
snr = "0001"
stype = "gy"
audience = "sus"      
ubb = False                   
ganztag = False
has_N = ["sus", "elt"]  
```

```python
meta_repo = MetaRepository()
meta_templates = meta_repo.meta_templates
meta_reports = meta_repo.meta_reports
meta_snames = meta_repo.meta_snames


meta_sets = meta_repo.meta_sets
```

```python
sname_meta = get_sname(
    meta_snames=meta_snames,
    snr = snr
    )

print("\n Report for:", sname_meta)
```

```python
limer_connect()
sids_df = limer_SIDs(snr=snr, ubb=ubb)
sid = sids_df["sid"]
print("\nReport template:", sid[0])
```

```python
sids_df.head()
```

```python
print(sids_df.sid)
```

```python
df = limer_responses(
    iSurveyID= sid[0],
    sCompletionStatus="complete"
)
    
limer_release()

print("\nData downloaded:", df.head())
```

```python
from pyreporter.utils import get_data
realdf = get_data(id = "251539", surveyls_title = "Bla", ubb = False)
realdf.head()
```
